```python
import numpy as np 
import pandas as pd 
```

```python
credits = pd.read_csv("tmdb_5000_credits.csv")
credits.head()
movies = pd.read_csv("tmdb_5000_movies.csv")
movies.head()
```

```python
credits.rename(
    columns={"movie_id": "id"},
    inplace=True
)
```

```python
movies = movies.merge(credits, on="id")
```

```python
movies = movies[
    ["id", "title_x", "genres", "keywords",
     "overview", "cast", "crew"]
].copy()
movies.rename(
    columns={"title_x": "title"},
    inplace=True
)
```

```python
movies.dropna(
    subset=["title", "overview", "genres", "keywords"],
    inplace=True
)
```

```python
import ast

def extract_names(text):
    data = ast.literal_eval(text)
    return [item["name"] for item in data]
```

```python
movies["genres"] = movies["genres"].apply(extract_names)

movies["keywords"] = movies["keywords"].apply(extract_names)
```

```python
def extract_cast(text):
    data = ast.literal_eval(text)
    return [item["name"] for item in data[:3]]
```

```python
movies["cast"] = movies["cast"].apply(extract_cast)
```

```python
print(movies[["title", "cast"]].head())
```

```python
def extract_director(text):
    data = ast.literal_eval(text)

    for item in data:
        if item["job"] == "Director":
            return [item["name"]]

    return []
```

```python
movies["crew"] = movies["crew"].apply(extract_director)
```

```python
movies["overview"] = movies["overview"].apply(
    lambda x: x.split()
)
```

```python
movies.head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.neighbors import NearestNeighbors
from sklearn.metrics.pairwise import cosine_similarity
import time
```

```python
print(movies.columns.tolist())
```

```python
movies["tags"] = movies.apply(
    lambda row: (
        row["genres"]
        + row["keywords"]
        + row["cast"]
        + row["crew"]
        + row["overview"]
    ),
    axis=1
)
```

```python
movies["tags"] = movies["tags"].apply(
    lambda x: " ".join(map(str, x))
)
```

```python
print(movies[["title", "tags"]].head())
```

```python
from sklearn.feature_extraction.text import CountVectorizer

count_vectorizer = CountVectorizer(
    max_features=5000,
    stop_words="english"
)

count_vectors = count_vectorizer.fit_transform(movies["tags"])

print("Count Vector Shape:", count_vectors.shape)
```

```python
from sklearn.neighbors import NearestNeighbors

knn_model = NearestNeighbors(
    n_neighbors=6,
    metric="cosine",
    algorithm="brute"
)

knn_model.fit(count_vectors)

print("KNN model ready!")
```

```python
def recommend_knn(movie_name):
    matches = movies[
        movies["title"].str.lower() == movie_name.lower()
    ]

    if matches.empty:
        print("Movie not found!")
        return

    index = matches.index[0]

    distances, indices = knn_model.kneighbors(
        count_vectors[index],
        n_neighbors=6
    )

    print(f"\nRecommendations for {movie_name}:\n")

    count = 0

    for movie_index, distance in zip(indices[0], distances[0]):
        if movie_index == index:
            continue

        title = movies.iloc[movie_index]["title"]
        similarity_score = 1 - distance

        print(f"{title} | Similarity: {similarity_score:.3f}")

        count += 1

        if count == 5:
            break
```

```python
recommend_knn("Avatar")
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import time

start_time = time.perf_counter()

tfidf = TfidfVectorizer(
    max_features=5000,
    stop_words="english"
)

tfidf_vectors = tfidf.fit_transform(movies["tags"])

tfidf_time = time.perf_counter() - start_time

print("TF-IDF Vector Shape:", tfidf_vectors.shape)
print("TF-IDF Processing Time:", round(tfidf_time, 3), "seconds")
```

```python
from sklearn.neighbors import NearestNeighbors

tfidf_knn = NearestNeighbors(
    n_neighbors=6,
    metric="cosine",
    algorithm="brute"
)

tfidf_knn.fit(tfidf_vectors)

print("TF-IDF + KNN model ready!")
```

```python
def recommend_tfidf(movie_name):
    matches = movies[movies["title"].str.lower() == movie_name.lower()]

    if matches.empty:
        print("Movie not found!")
        return

    index = matches.index[0]

    distances, indices = tfidf_knn.kneighbors(
        tfidf_vectors[index],
        n_neighbors=6
    )

    print(f"\nTF-IDF Recommendations for {movie_name}:\n")

    count = 0

    for movie_index, distance in zip(indices[0], distances[0]):
        if movie_index == index:
            continue

        title = movies.iloc[movie_index]["title"]
        similarity_score = 1 - distance

        print(f"{title} | Similarity: {similarity_score:.3f}")

        count += 1
        if count == 5:
            break
```

```python
recommend_tfidf("Avatar")
```

```python
print("=" * 50)
recommend_knn("Avatar")

print("\n" + "=" * 50)
recommend_tfidf("Avatar")
```

```python
import time

# CountVectorizer + KNN
start = time.perf_counter()
recommend_knn("Avatar")
count_time = time.perf_counter() - start

# TF-IDF + KNN
start = time.perf_counter()
recommend_tfidf("Avatar")
tfidf_recommendation_time = time.perf_counter() - start

print("\n========== EXECUTION TIME ==========")
print(f"CountVectorizer + KNN: {count_time:.6f} seconds")
print(f"TF-IDF + KNN: {tfidf_recommendation_time:.6f} seconds")
```

```python
mood_genres = {
    "Happy 😊": ["Comedy", "Family", "Animation"],
    "Sad 😢": ["Drama", "Romance"],
    "Stressed 😌": ["Comedy", "Family", "Animation"],
    "Adventurous 🤩": ["Adventure", "Action", "Fantasy"],
    "Thrilled 😱": ["Thriller", "Mystery", "Crime"],
    "Romantic ❤️": ["Romance", "Comedy"],
    "Scared 👻": ["Horror", "Thriller"],
    "Curious 🧠": ["Mystery", "Science Fiction", "Documentary"]
}

print("Available moods:\n")

for mood, genres in mood_genres.items():
    print(f"{mood}: {', '.join(genres)}")
```

```python
import random

def recommend_by_mood(mood, number=5):
    if mood not in mood_genres:
        print("Mood not available!")
        return

    selected_genres = mood_genres[mood]

    # Movies matching the selected mood's genres
    recommendations = movies[
        movies["genres"].apply(
            lambda genres: any(
                genre in genres for genre in selected_genres
            )
        )
    ].copy()

    if recommendations.empty:
        print("No movies found for this mood!")
        return

    # Randomly select movies
    number = min(number, len(recommendations))
    recommendations = recommendations.sample(
        n=number,
        random_state=None
    )

    print(f"\n🎬 Movies for mood: {mood}\n")

    for _, movie in recommendations.iterrows():
        print(f"🎥 {movie['title']}")
        print(f"Genres: {', '.join(movie['genres'])}")
        print("-" * 40)
```

```python
recommend_by_mood("Happy 😊", 5)
```

```python
recommend_by_mood("Stressed 😌", 5)
```

```python
mood_tags = {
    "Happy 😊": "happy funny comedy feel good friendship family fun",
    "Sad 😢": "emotional drama loss struggle hope life",
    "Stressed 😌": "comedy relaxing friendship family feel good",
    "Adventurous 🤩": "adventure action journey exploration fantasy",
    "Thrilled 😱": "thriller mystery suspense crime investigation",
    "Romantic ❤️": "romance love relationship romantic comedy",
    "Scared 👻": "horror fear supernatural ghost suspense",
    "Curious 🧠": "mystery science fiction discovery investigation"
}

print("Mood reference tags ready!")
```

```python
from sklearn.metrics.pairwise import cosine_similarity

def recommend_mood_hybrid(mood, number=5):

    if mood not in mood_genres:
        print("Mood not available!")
        return

    # Step 1: Filter movies by mood-related genres
    selected_genres = mood_genres[mood]

    candidates = movies[
        movies["genres"].apply(
            lambda genres: any(
                genre in genres for genre in selected_genres
            )
        )
    ].copy()

    if candidates.empty:
        print("No matching movies found!")
        return

    # Step 2: Convert mood description into TF-IDF features
    mood_vector = tfidf.transform([mood_tags[mood]])

    # Step 3: Calculate similarity with candidate movies
    candidate_vectors = tfidf_vectors[candidates.index]

    scores = cosine_similarity(
        mood_vector,
        candidate_vectors
    ).flatten()

    # Step 4: Rank movies by similarity
    candidates["mood_score"] = scores

    candidates = candidates.sort_values(
        by="mood_score",
        ascending=False
    )

    # Step 5: Display top recommendations
    results = candidates.head(number)

    print(f"\n🎬 Recommendations for {mood}\n")

    for _, movie in results.iterrows():
        print(f"🎥 {movie['title']}")
        print(f"Genres: {', '.join(movie['genres'])}")
        print(f"Mood similarity: {movie['mood_score']:.3f}")
        print("-" * 40)
```

```python
recommend_mood_hybrid("Stressed 😌", 5)
```
